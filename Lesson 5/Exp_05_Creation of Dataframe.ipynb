```python
import pandas as pd 
```

```python
#creating ones one dataframe
#pd.DataFrame(): Creates a 2D table (like Excel sheet)

#First argument: List of lists = rows of data

#columns: Names for each column

df = pd.DataFrame([[10,15,20,16],[19,17,15,13],[15,19,17,14]],
                  columns=["CD","DBMS","DSS","CAO"])
```

```python
#pd.DataFrame(): Creates a 2D table (like Excel sheet)

#First argument: List of lists = rows of data

#columns: Names for each column

df #visualization of data frame 
```

```python
df.shape    # Returns (3, 4) → (rows, columns)
```

```python
df.size     # Returns 12 → total cells (3×4)
```

```python
df.ndim     # Returns 2 → 2-dimensional table
```

```python


# Create new DataFrame with 1 row

df2=pd.DataFrame([[5,6,7,8]],
                columns=["CD","DBMS","DSS","CAO"])
```

```python
df2
```

```python
# Append with reset index
df3=pd.concat([df, df2], ignore_index=True)
```

```python
df3
```

```python
df3["DM"] = [12,15,17,19]  # Add new column "DM" (Data Mining)
```

```python
df3
```

```python
# Delete ROWS by index
df4 = df3.drop(index=[1])  # Remove row with index 1
```

```python
df4
```

```python
# Delete COLUMNS by name  
df5 = df3.drop(columns=["DM"])  # Remove "DM" column
```

```python
df5
```

```python
#finding mean of DSS
print("Mean of DSS:",df5["DSS"].mean())
```

```python
#finding median of DSS
print("Median of DSS:",df5["DSS"].median())
```

```python
#finding mode of DSS,Most frequent: all appear once

print("Mode of DSS:",df5["DSS"].mode())
```

```python
#finding max of DSS
print("Max of DSS:",df5["DSS"].max())
```

```python
#finding min of DSS
print("Min of DSS:",df5["DSS"].min())
```
